## Sub-gráficos

```python
df = pd.read_csv("../data/census.csv")
df.info()
```

```python
df.plot(subplots=True);
```

```python
df.plot(
    subplots=True, layout=(3,3), figsize=(10,10),
    title=['Age', 'Final', 'Education', 'Gain', 'Loss', 'Hous'],
    legend=False, kind='hist'
);
```

## Gráfico de Barras

```python
df.workclass.unique()
```

```python
df.groupby("workclass")["age"].count().plot(kind="bar");
```

```python
df.groupby("workclass")["age"].count().plot(kind="barh");
```

```python
pivot = df.query("workclass != ' Never-worked' and workclass != ' Without-pay'").pivot_table(
    index='income', columns='workclass', values='education-num', aggfunc='sum'
).apply(lambda x: x * 100 / sum(x), axis=1)
pivot
```

```python
pivot.plot.bar().legend(bbox_to_anchor=(1.4,1));
```

```python
pivot.plot.bar(stacked=True).legend(bbox_to_anchor=(1.4,1));
```

## Gráfico de Pizza

```python
df.groupby("workclass")["age"].count().plot.pie(title='worklass', ylabel="");
```
